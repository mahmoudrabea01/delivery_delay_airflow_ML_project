to see the data files

```python
from pathlib import Path

# Define the project directory
PROJECT_DIR = Path.cwd().parent

# Define the raw data directory
RAW_DATA_DIR = PROJECT_DIR / "data" / "raw"

# Find all CSV files
csv_files = sorted(RAW_DATA_DIR.glob("*.csv"))

print(f"Raw data directory: {RAW_DATA_DIR}")
print(f"Number of CSV files: {len(csv_files)}")

for file in csv_files:
    print(file.name)
```

read and merage the first three .csv

```python
import pandas as pd

first_three_files = csv_files[:3]
orders = pd.concat(
    [pd.read_csv(file_path) for file_path in first_three_files],
    ignore_index=True,
)

print("Files loaded:")
for file_path in first_three_files:
    print(file_path.name)

print(f"Combined shape: {orders.shape}")
orders.head()
```
